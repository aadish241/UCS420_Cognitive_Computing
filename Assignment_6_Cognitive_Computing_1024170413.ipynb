## Q1. Sensor Readings and Vectorization

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add +2°C to every reading using vectorization
corrected_temperature = temperature + 2
print("Original readings:", temperature)
print("After adding +2°C:", corrected_temperature)

# b) Convert Celsius to Fahrenheit: F = (9/5)C + 32
fahrenheit = (9/5) * corrected_temperature + 32
print("\nFahrenheit readings:", fahrenheit)

# c) Identify readings greater than 32°C using Boolean indexing
readings_greater_32 = corrected_temperature[corrected_temperature > 32]
print("\nReadings greater than 32°C:", readings_greater_32)

# d) Count readings exceeding 32°C
count_greater_32 = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count_greater_32)
```

### Q1(e) Explanation
**Vectorization** applies an operation to an entire NumPy array at once, avoiding an explicit Python `for` loop. NumPy performs these operations efficiently at a lower level, making the code shorter and generally faster for large datasets.

**Boolean indexing** creates a True/False mask such as `corrected_temperature > 32` and directly selects the values satisfying the condition. This is more concise and efficient than manually checking every reading.

## Q2. Daily Steps Matrix

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix:")
print(steps)

# a) Total steps
print("\na) Total steps:", np.sum(steps))

# b) Mean number of steps
print("b) Mean steps:", np.mean(steps))

# c) Maximum and minimum
print("c) Maximum:", np.max(steps))
print("   Minimum:", np.min(steps))

# d) Total steps for each day (axis=0)
print("d) Total for each day:", np.sum(steps, axis=0))

# e) Total steps for each user (axis=1)
print("e) Total for each user:", np.sum(steps, axis=1))

# f) Position of maximum value using argmax()
flat_index = np.argmax(steps)
position = np.unravel_index(flat_index, steps.shape)

print("f) Maximum value:", steps[position])
print("   Position (0-based user, day):", position)
print("   Position (1-based user, day):",
      (position[0] + 1, position[1] + 1))
```

## Q3. NumPy Slicing, Copying, Reshaping and Flattening

```python
# a) Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("a) original:", original)

# b) Slice elements from index 1 to 4 (index 4 excluded)
subset = original[1:4]
print("\nb) subset:", subset)

# c) Modify first element of subset
subset[0] = 999
print("\nc) After subset[0] = 999:")
print("original:", original)
print("subset:", subset)
print("The original changes because a slice is a view of the original array.")

# d) Use .copy()
copied_array = original[1:4].copy()
copied_array[0] = 500

print("\nd) After copied_array[0] = 500:")
print("original:", original)
print("copied array:", copied_array)
print("The original does not change because .copy() creates an independent array.")

# e) Create 1 to 12 and reshape to 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)
print("\ne) 3 x 4 matrix:")
print(matrix)
```

```python
# f) Extract requested rows, columns and submatrix

first_row = matrix[0, :]
last_row = matrix[-1, :]
second_column = matrix[:, 1]

# Rows 1–2 and columns 2–3, using the question's 1-based wording
# NumPy indices are 0-based, so this is matrix[0:2, 1:3].
rows_1_2_cols_2_3 = matrix[0:2, 1:3]

print("f) First row:", first_row)
print("Last row:", last_row)
print("Second column:", second_column)
print("Rows 1–2, columns 2–3:")
print(rows_1_2_cols_2_3)
```

```python
# g) Flatten using flatten() and ravel()
flattened_array = matrix.flatten()
raveled_array = matrix.ravel()

print("g) flatten():", flattened_array)
print("ravel():", raveled_array)

# h) Modify ravel() result
raveled_array[0] = 9999
print("\nh) After changing raveled_array[0] to 9999:")
print("raveled_array:", raveled_array)
print("matrix:")
print(matrix)
print("ravel() usually returns a view when possible, so the original is affected.")

# i) Modify flatten() result
flattened_array[1] = 8888
print("\ni) After changing flattened_array[1] to 8888:")
print("flattened_array:", flattened_array)
print("matrix:")
print(matrix)
print("flatten() returns a copy, so the original is not affected.")
```

```python
# j) Shape, dimensions, size and data type
print("j) Shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:", matrix.size)
print("dtype:", matrix.dtype)
```

## Q4. Cognitive-Assistive System and Ordinary Least Squares

```python
y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# a) Shape and dimensions
print("a) Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b) Transpose
XT = X.T
print("\nb) X.T:")
print(XT)

print("\nX.T changes X from 4x3 to 3x4.")
print("Each row of X.T represents one feature:")
print("Row 1: Sleep hours")
print("Row 2: Activity level")
print("Row 3: Stress level")
```

```python
# c) Matrix product X.T @ X
XTX = X.T @ X
print("c) X.T @ X:")
print(XTX)

# d) Matrix inverse
XTX_inverse = np.linalg.inv(XTX)
print("\nd) Inverse of X.T @ X:")
print(XTX_inverse)
```

```python
# e) Ordinary Least Squares:
# beta = (X.T X)^(-1) X.T y
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("e) OLS coefficients (beta):")
print(beta)

# Equivalent approach using solve()
beta_check = np.linalg.solve(X.T @ X, X.T @ y)
print("\nVerification using np.linalg.solve():")
print(beta_check)
```

### Q4(f) Interpretation of coefficients
The three coefficients correspond respectively to **sleep hours, activity level, and stress level**.

Each coefficient represents the estimated change in the assistance score for a one-unit increase in that feature while the other features remain constant.

- A **positive** coefficient indicates a positive association with the predicted assistance score.
- A **negative** coefficient indicates a negative association.
- The supplied OLS equation has **no intercept**, so these are the coefficients of the three given features only.

```python
# g) Predict the assistance score for a new user
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("g) New user:", new_user)
print("Predicted assistance score:", predicted_score)
```
