## Q1. Basic Operations on a NumPy 1-D Array

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

## Q2. Basic NumPy Array Questions
### (a) Reverse the NumPy array

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

### (b) Find the most frequent value and its indices

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Array x:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", counts[np.argmax(counts)])
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Array y:", y)
print("Most frequent value:", most_frequent)
print("Frequency:", counts[np.argmax(counts)])
print("Indices:", indices)
```

## Q3. Access Elements Using Row and Column Indices

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Array:")
print(arr)

# NumPy uses zero-based indexing
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

## Q4. NumPy Array Using linspace()

```python
Aadish = np.linspace(10, 100, 25)

print("Aadish array:")
print(Aadish)
print("\nDimensions:", Aadish.ndim)
print("Shape:", Aadish.shape)
print("Total elements:", Aadish.size)
print("Data type:", Aadish.dtype)
print("Total bytes consumed:", Aadish.nbytes)

# Transpose using reshape()
Aadish_transpose = Aadish.reshape(25, 1).T
print("\nTranspose using reshape():")
print(Aadish_transpose)

# T attribute
Aadish_T = Aadish.reshape(25, 1).T
print("\nTranspose using T attribute:")
print(Aadish_T)

print("\nAre both results equal?", np.array_equal(Aadish_transpose, Aadish_T))
print("Direct Aadish.T shape:", Aadish.T.shape)
print("For a 1-D array, .T does not change its shape; reshape() is needed first.")
```

## Q5. 2-D Array Operations

```python
ucs420_Aadish = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Aadish)

print("\nMean:", np.mean(ucs420_Aadish))
print("Median:", np.median(ucs420_Aadish))
print("Maximum:", np.max(ucs420_Aadish))
print("Minimum:", np.min(ucs420_Aadish))
print("Unique elements:", np.unique(ucs420_Aadish))
```

```python
reshaped_ucs420_Aadish = ucs420_Aadish.reshape(4, 3)

print("reshaped_ucs420_Aadish:")
print(reshaped_ucs420_Aadish)
```

```python
resized_ucs420_Aadish = np.resize(ucs420_Aadish, (2, 3))

print("resized_ucs420_Aadish:")
print(resized_ucs420_Aadish)
```
